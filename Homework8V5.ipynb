Question 1 (35 pt)

A sample of 30 respondents was interviewed using mall intercept interviewing. The respondents were asked to indicate their degree of agreement with the following statements using a seven-point scale (1 = strongly disagree, 7 = strongly agree).

вЂў	V1 = It is important to buy a toothpaste that prevents cavities

вЂў	V2 = I like a toothpaste that gives a shiny teeth

вЂў	V3 = A toothpaste should strengthen your gums teeth

вЂў	V4 = I prefer a toothpaste that freshens breath

вЂў	V5 = Prevention of tooth decay is not an important benefit offered by a toothpaste

вЂў	V6 = The most important consideration in buying a toothpaste is attractive teeth

By using the variables,

(1) How many factors is extracted from data for principal component analysis?

(2) What is the total percentage of variance explained by these principal components?

(3)Which variables are included in the same factors ? How can you name these factors?

You can find the data at

https://raw.githubusercontent.com/ogut77/DataScience/main/data/Toothpaste.csv

```python
# Q1 - Principal Component Analysis / Factor Analysis for toothpaste data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

tooth = pd.read_csv('https://raw.githubusercontent.com/ogut77/DataScience/main/data/Toothpaste.csv')
display(tooth.head())

X_tooth = tooth[['V1', 'V2', 'V3', 'V4', 'V5', 'V6']]
X_scaled = StandardScaler().fit_transform(X_tooth)

pca = PCA()
pca.fit(X_scaled)
eigenvalues = pca.explained_variance_
explained = pca.explained_variance_ratio_

pca_summary = pd.DataFrame({
    'Component': [f'PC{i+1}' for i in range(len(eigenvalues))],
    'Eigenvalue': eigenvalues,
    'Variance Explained': explained,
    'Cumulative Variance': explained.cumsum()
})
display(pca_summary)

n_components = int((eigenvalues > 1).sum())
print('Number of components with eigenvalue > 1:', n_components)
print('Total variance explained by selected components:', round(explained[:n_components].sum() * 100, 2), '%')

loadings = pd.DataFrame(
    pca.components_[:n_components].T,
    index=X_tooth.columns,
    columns=[f'PC{i+1}' for i in range(n_components)]
)
display(loadings)

for component in loadings.columns:
    vars_in_factor = loadings.index[loadings[component].abs() >= 0.40].tolist()
    print(component, 'variables:', vars_in_factor)

plt.figure(figsize=(7, 4))
plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, marker='o')
plt.axhline(1, color='red', linestyle='--')
plt.xlabel('Principal Component')
plt.ylabel('Eigenvalue')
plt.title('Scree Plot')
plt.show()

print('Suggested interpretation: variables with large absolute loadings on the same component belong to the same factor.')
print('A reasonable naming is based on the grouped variables, such as health/cavity prevention and appearance/fresh breath benefits.')
```

```python
 #There are missing values in the Lead data .
# Replace missing variables with the mean  if variable is numeric or mode if variable is categorical.
for col in df.columns:
    if ((df[col].dtype == 'float64') or (df[col].dtype == 'float64')) :
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Check that there are no missing data
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

df_model = df.copy()
for col in df_model.select_dtypes(include='object').columns:
    le = LabelEncoder()
    df_model[col] = le.fit_transform(df_model[col].astype(str))

df_model = df_model.drop(columns=['Prospect ID','Lead Number'])
y = df_model['Converted']
X = df_model.drop(columns=['Converted'])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

train_data = X_train.copy()
train_data['Converted'] = y_train
test_data = X_test.copy()
test_data['Converted'] = y_test

print(X_train.shape, X_test.shape)
```

```python
def eval(yact,ypred):
 from sklearn import metrics
 print("Accuracy:",metrics.accuracy_score(yact, ypred))
 print("Precision:",metrics.precision_score(yact, ypred))
 print("Recall:",metrics.recall_score(yact, ypred))

 cnf_matrix = metrics.confusion_matrix(yact, ypred)
 import numpy as np
 import matplotlib.pyplot as plt
 import seaborn as sns
 %matplotlib inline
 class_names=[0,1] # name  of classes
 fig, ax = plt.subplots()
 tick_marks = np.arange(len(class_names))
 plt.xticks(tick_marks, class_names)
 plt.yticks(tick_marks, class_names)
 # create heatmap
 sns.heatmap(pd.DataFrame(cnf_matrix), annot=True, cmap="YlGnBu" ,fmt='g')
 ax.xaxis.set_label_position("top")
 plt.tight_layout()
 plt.title('Confusion matrix', y=1.1)
 plt.ylabel('Actual label')
 plt.xlabel('Predicted label')
```

```python
#Q3(35 pt)-Using data below, find how many clusters are there based on elbow method for k-means
#and hierarchical clustering
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

cl = pd.read_csv('https://raw.githubusercontent.com/ogut77/DataScience/refs/heads/main/cluster.csv')
display(cl.head())

X_cl = cl.select_dtypes(include=np.number)
X_cl_scaled = StandardScaler().fit_transform(X_cl)

inertias = []
k_values = range(1, 11)
for k in k_values:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_cl_scaled)
    inertias.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(list(k_values), inertias, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for K-Means')
plt.xticks(list(k_values))
plt.show()

# Estimate elbow by maximum distance from the line between k=1 and k=10.
points = np.column_stack((list(k_values), inertias))
line_start, line_end = points[0], points[-1]
distances = np.abs(np.cross(line_end - line_start, line_start - points)) / np.linalg.norm(line_end - line_start)
best_k = int(list(k_values)[int(np.argmax(distances))])
print('Estimated best k from elbow method:', best_k)

Z = linkage(X_cl_scaled, method='ward')
plt.figure(figsize=(10, 5))
dendrogram(Z)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Observations')
plt.ylabel('Distance')
plt.show()

hier_model = AgglomerativeClustering(n_clusters=best_k, linkage='ward')
cl['KMeansCluster'] = KMeans(n_clusters=best_k, random_state=42, n_init=10).fit_predict(X_cl_scaled)
cl['HierarchicalCluster'] = hier_model.fit_predict(X_cl_scaled)
display(cl.head())
print('Based on the elbow plot and dendrogram, the selected number of clusters is approximately:', best_k)
```
